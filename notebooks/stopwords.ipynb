```python
import nltk
from nltk.corpus import stopwords
import pandas as pd # Import pandas
nltk.download('stopwords')
stop_words=stopwords.words('english')
more_stopwords=['u','im','c']
stop_words=stop_words+more_stopwords

def remove_stopwords(text):
  # Ensure text is treated as a string, especially if there are non-string elements
  if not isinstance(text, str):
    return '' # Or handle as appropriate, e.g., str(text)
  text=' '.join(word for word in text.split(' ') if word not in stop_words)
  return text

# Placeholder for df definition - replace with your actual data loading
# Ensure df is a pandas DataFrame; if not, create a placeholder DataFrame.
if not ('df' in globals() and isinstance(globals()['df'], pd.DataFrame)):
    data = {'message_clean': ['This is a sample message with u and im words.', 'Another example message c for testing.']}
    df = pd.DataFrame(data)

df['message_clean']=df['message_clean'].apply(remove_stopwords)
df.head()
```
